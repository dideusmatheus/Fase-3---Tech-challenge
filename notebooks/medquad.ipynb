```python
df=df.dropna()
df.head()
```

```python
df = df.reset_index()
df.head()
```

```python
df=df.drop('index',axis=1)
df.head()
```

```python
def extract_text(focus_area):
    text=""
    data=df[df['focus_area']==focus_area]
    for i in data['answer']:
        text+=i
    return text
```

```python
from langchain_core.documents import Document
```

```python
df['focus_area'].value_counts()
```

```python
documented_text=[]
for i in df['focus_area'].unique():
    documented_text.append(Document(page_content=extract_text(i),metadata={"focus_area":i}))
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=3000,chunk_overlap=300)
```

```python
chunked_data = splitter.split_documents(documented_text)
```
